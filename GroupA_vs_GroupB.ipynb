```python
# ============================================================
# Group A vs Group B – Statistics & Outlier Detection
# ============================================================

# Import required libraries
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Create the datasets
# ------------------------------------------------------------

Group_A = [45, 48, 50, 52, 49, 51, 47, 50]
Group_B = [60, 62, 65, 63, 61, 64, 66, 95]


# ------------------------------------------------------------
# 2. Function to calculate statistics
# ------------------------------------------------------------

def calculate_statistics(data, group_name):

    print("\n====================================")
    print(group_name)
    print("====================================")

    # Mean
    mean = np.mean(data)

    # Median
    median = np.median(data)

    # Minimum
    minimum = np.min(data)

    # Maximum
    maximum = np.max(data)

    # Standard Deviation
    std = np.std(data)

    # Quartiles
    Q1 = np.percentile(data, 25)
    Q3 = np.percentile(data, 75)

    # Interquartile Range
    IQR = Q3 - Q1

    # Lower and upper limits for outliers
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    # Find outliers
    outliers = [
        x for x in data
        if x < lower_limit or x > upper_limit
    ]

    # Display results
    print("Mean              :", mean)
    print("Median            :", median)
    print("Minimum           :", minimum)
    print("Maximum           :", maximum)
    print("Standard Deviation:", round(std, 2))
    print("Q1                :", Q1)
    print("Q3                :", Q3)
    print("IQR               :", IQR)
    print("Lower Limit")
```
