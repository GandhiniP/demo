```python
import pandas as pd
df=pd.read_excel("FDP17data.xlsx")
df
```

```python
df.head()
print(df.shape)
df.columns
```

```python
print(df.groupby("Department")["Marks"].max())
print(df.groupby("Department")["Marks"].min())
```

```python

print(df.groupby("Department")["Marks"].agg(["count","max","min","median"]))
```

```python
print(df.groupby("Department")[["Marks","Attendance_Percent"]].mean())
```

```python
pd.pivot_table(
    df,
    index="Department",
    values="Marks",
    columns="Class",
    aggfunc="mean"
)
```

```python
pd.crosstab(df["Department"],df["Result"])
```
