```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/CAR DETAILS FROM CAR DEKHO.csv')
print(df.head())
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(y=df['selling_price'])

plt.title('Box Plot of Car Selling Price')
plt.ylabel('Selling Price')
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['year'], df['selling_price'])

plt.title('Year vs Selling Price')
plt.xlabel('Year')
plt.ylabel('Selling Price')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['selling_price'], bins=20, edgecolor='black')

plt.title('Distribution of Car Selling Price')
plt.xlabel('Selling Price')
plt.ylabel('Number of Cars')

plt.show()
```

```python
fuel_count = df['fuel'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    fuel_count,
    labels=fuel_count.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Distribution of Cars by Fuel Type')
plt.show()
```

```python
fuel_count = df['fuel'].value_counts()

plt.figure(figsize=(8,5))

plt.bar(fuel_count.index, fuel_count.values)

plt.title('Number of Cars by Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('Number of Cars')

plt.show()
```

```python
year_price = df.groupby('year')['selling_price'].mean()

plt.figure(figsize=(10,5))

plt.plot(year_price.index, year_price.values, marker='o')

plt.title('Average Car Selling Price by Year')
plt.xlabel('Year')
plt.ylabel('Average Selling Price')

plt.grid()
plt.show()
```
